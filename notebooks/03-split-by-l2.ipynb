```python
import pandas as pd
import re
```

```python
data_processed = pd.read_csv("../data/03-encoded-properati.csv", sep=',', index_col=0)
data_processed.head(5)
```

```python
def to_snake_case(name: str) -> str:
    name = re.sub(r"[^\w]", "_", name)        # replace spaces & symbols with _
    name = re.sub(r"([a-z0-9])([A-Z])", r"\1_\2", name)
    name = re.sub(r"_+", "_", name)            # collapse multiple _
    return name.lower().strip("_")

l2_dummy_cols = [c for c in data_processed.columns if c.startswith("l2_")]

for col in l2_dummy_cols:
    data_processed[data_processed[col] == 1].to_csv("../data/split_by_l2/" + to_snake_case(col) + ".csv")

rows_all_zero = data_processed[data_processed[l2_dummy_cols].eq(0).all(axis=1)]
rows_all_zero.to_csv("../data/split_by_l2/l2_others.csv")
```
